# Dataset exploration
Counts per defect type and defect-size distribution (small defects drive Gap 2).

```python
import sys; sys.path.insert(0, "..")
from src.utils.config import load_config, resolve_path
cfg = load_config()
CATEGORY = "bottle"   # change me
import numpy as np, matplotlib.pyplot as plt
from collections import Counter
from src.data.dataloader import build_datasets
from src.data.preprocessing import Preprocessor
pre = Preprocessor.from_config(cfg["data"]); ds = build_datasets(cfg, CATEGORY, pre)
print({k: len(v) for k, v in ds.items()})
print(Counter(s["defect_type"] for s in ds["test"].samples))
```

```python
areas = [float(ds["test"][i]["mask"].mean()) for i in range(len(ds["test"])) if ds["test"][i]["label"]]
plt.hist(100*np.array(areas), bins=20); plt.xlabel("defect area (% of image)"); plt.ylabel("count"); plt.show()
print("median %%: %.3f   smallest %%: %.4f" % (100*np.median(areas), 100*min(areas)))
```

```python
fig, ax = plt.subplots(2, 4, figsize=(12, 6))
for a, i in zip(ax.ravel(), np.linspace(0, len(ds["test"])-1, 8).astype(int)):
    s = ds["test"][i]; a.imshow(pre.denormalize(s["image"])); a.set_title(s["defect_type"], fontsize=8); a.axis("off")
plt.show()
```
