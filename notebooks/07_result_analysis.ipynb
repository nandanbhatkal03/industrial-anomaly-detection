# Result analysis
Reads `results/tables/summary.csv`; per-category and per-defect-type analysis, trade-off plot.

```python
import sys; sys.path.insert(0, "..")
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv("../results/tables/summary.csv")
cols = ["image_auroc","pixel_auroc","aupro","image_f1","pixel_iou","latency_ms_mean","fps","params_m","deployed_size_mb"]
display(df.groupby(["method","tag"])[cols].mean().round(4))
display(df.pivot_table(index="category", columns="method", values="aupro").round(3))
```

```python
g = df.groupby(["method","tag"])[["latency_ms_mean","image_auroc","deployed_size_mb"]].mean().reset_index()
plt.scatter(g.latency_ms_mean, g.image_auroc, s=40+4*g.deployed_size_mb)
for _, r in g.iterrows(): plt.annotate(f"{r.method}:{r.tag}", (r.latency_ms_mean, r.image_auroc), fontsize=7)
plt.xlabel("latency (ms)"); plt.ylabel("image AUROC"); plt.show()
```
