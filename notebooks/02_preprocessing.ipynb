# Preprocessing study
Does resizing / denoising erase small defects? Compares surviving defect area and sharpness.

```python
import sys; sys.path.insert(0, "..")
from src.utils.config import load_config, resolve_path
cfg = load_config()
CATEGORY = "bottle"   # change me
import numpy as np
from src.data.dataset import MVTecDataset
from src.data.preprocessing import Preprocessor
root = resolve_path(cfg["paths"]["data_root"])
for size in (128, 224, 256, 384):
    for dn in ("none", "gaussian", "nlm"):
        pre = Preprocessor(size, None, dn)
        ds = MVTecDataset(root, CATEGORY, "test", pre)
        bad = [ds[i] for i in range(len(ds)) if ds[i]["label"]]
        px = np.mean([float(s["mask"].sum()) for s in bad])
        print(f"size={size:4d} denoise={dn:9s} mean defect pixels={px:8.1f}")
```
